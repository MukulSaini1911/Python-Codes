## String Formatting in Python

```python
def greeting(name):
    return "Hello {} Welcome to France.".format(name)
```

```python
greeting('Mukul')
```

```python
def greeting(name, age):
    return "Hello {}. Your age is {}.".format(name, age)
```

```python
greeting('Mukul', 24)
```

```python
def greeting(name, age):
    return "Hello {name1}. Your age is {age1}.".format(age1=age, name1=name)
```

```python
greeting('Mukul', 24)
```

## Python List iterables vs iterators

```python
## List is iterable

lst = [1,2,3,4,5]
for i in lst:
    print(i)
```

```python
## iterator
lst1 = iter(lst)
```

```python
# show the memory allocation
lst1
```

```python
# to handle the StopIteration issue, we can use the for loop to iterate over all the values available in the lst1 = iter(lst)
lst1 = iter(lst)
for i in lst1:
    print(i)
```
