```python
class Car:
    def __init__(self,window,door,engine_type):  # Consturctor
        self.windows = window                    # self is nothing but a reference to the perticular object.
        self.doors = door
        self.engine_type = engine_type
    def self_driving(self):                      # method creation
        return "This is a {} car.".format(self.engine_type)
```

```python
car1 = Car(3,3,'diesel')
```

```python
car1.self_driving()
```

```python
car1.doors
```

```python
car1.engine_type
```
